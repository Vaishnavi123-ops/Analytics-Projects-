```python
import pandas as pd
df= pd.read_csv("Telco-Customer-Churn2.csv")
df.head(5)
```

```python
df.info()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.dtypes
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['TotalCharges'])
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
del df['customerID']
```

```python
df.columns
```

```python
df[['Churn','Partner','Dependents','PhoneService','PaperlessBilling']]=df[['Churn','Partner','Dependents','PhoneService','PaperlessBilling']].replace({'Yes':1,'No':0})
```

```python
df.head(10)
```

```python
y = df['Churn']
X = df.drop('Churn', axis=1)
```

```python
X_Encoded = pd.get_dummies(X, drop_first=False)
```

```python
X_Encoded.head(10)
```

```python
df.columns
```

```python
final_df = pd.concat([X_Encoded, y], axis=1)
```

```python
final_df.head(10)
```

```python
final_df.dtypes
```

```python
final_df=final_df.astype(int)
```

```python
final_df.dtypes
```

```python
final_df.head(10)
```

```python
cols_to_drop = ['gender_Male',
'MultipleLines_No phone service',
'OnlineSecurity_No internet service',
'OnlineBackup_No internet service',
'DeviceProtection_No internet service',
'TechSupport_No internet service',
'StreamingTV_No internet service',
'StreamingMovies_No internet service',
'InternetService_No',

]


final_df = final_df.drop(columns=cols_to_drop, errors='ignore')
```
